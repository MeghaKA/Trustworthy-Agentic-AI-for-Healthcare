```python
import pandas as pd
import numpy as np
import os

os.makedirs("data/processed", exist_ok=True)
os.makedirs("data/raw", exist_ok=True)

pd.set_option("display.max_columns", None)
```

```python
from pathlib import Path

DATA_PATH = Path("data/raw")

files = [

    "P_DEMO.xpt",
    "P_BMX.xpt",
    "P_BPXO.xpt",
    "P_GLU.xpt",
    "P_GHB.xpt",
    "P_TCHOL.xpt",
    "P_TRIGLY.xpt",
    "P_HDL.xpt",
    "P_CBC.xpt",
    "P_BIOPRO.xpt",
    "P_ALB_CR.xpt",
    "P_IRON.xpt",
    "P_HSCRP.xpt",
    "P_DIQ.xpt",
    "P_DBQ.xpt",
    "P_PAQ.xpt",
    "P_SLQ.xpt",
    "P_ALQ.xpt",
    "P_SMQ.xpt",
    "P_HIQ.xpt",
    "P_HUQ.xpt",
    "P_RXQ_RX.xpt"

]
```

```python
datasets = {}

missing_files = []

for file in files:

    path = DATA_PATH / file

    if path.exists():

        datasets[file] = pd.read_sas(path)

        print(f"Loaded : {file}")

        print(f"Shape  : {datasets[file].shape}")

        print("------------------------")

    else:

        print(f"Missing : {file}")

        missing_files.append(file)
```

```python
print("Total datasets loaded :", len(datasets))

print()

print("Missing datasets :", len(missing_files))

missing_files
```

```python
master = datasets["P_DEMO.xpt"].copy()

for name, df in datasets.items():

    if name == "P_DEMO.xpt":

        continue

    master = master.merge(

        df,

        on="SEQN",

        how="left"

    )
```

```python
print(master.shape)

master.head()
```

```python
missing_percent = (

    master

    .isnull()

    .mean()

    *100

)

missing_report = pd.DataFrame({

    "Feature":master.columns,

    "Missing (%)":missing_percent.values

})

missing_report = missing_report.sort_values(

    "Missing (%)",

    ascending=False

)

missing_report.head(30)
```

```python
master = master.loc[

    :,

    ~master.columns.duplicated()

]
```

```python
master.to_csv(

    "data/processed/master_healthcare_dataset.csv",

    index=False

)

print("Dataset Saved Successfully")

print()

print(master.shape)
```
