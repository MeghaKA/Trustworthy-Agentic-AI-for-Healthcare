# NB9 — Automated Trustworthy Healthcare AI Demonstrator

## Project
**Trustworthy Multi-Agent Clinical Decision Support System using Explainable AI and NHANES Data**

## Stage
Stage 9 — Automated Research Prototype / Demonstrator

---

## 1. Research Objective

This notebook develops an automated research demonstrator that operationalizes the
governance-aware multi-agent healthcare workflow established in NB8.

The demonstrator integrates:

1. Input and Data Quality Validation
2. Locked Predictive Modelling
3. Patient-Level Model Explanation
4. Trust and Fairness Evidence
5. Mandatory Safety Gating
6. Structured Clinical Decision-Support Reporting
7. Agentic Orchestration
8. Provenance and Audit Logging

The demonstrator is designed as a research prototype rather than a clinical
diagnostic or treatment system.

---

## 2. Research Questions

### RQ1
Can the validated NB8 multi-agent healthcare workflow be operationalized as an
automated, reproducible research demonstrator?

### RQ2
Can model prediction, explanation, trust evidence and safety governance be
integrated into a single auditable workflow without modifying the locked
predictive model or decision threshold?

### RQ3
Can mandatory safety controls prevent clinically inappropriate interpretation
when patient input is incomplete or unsuitable for individualized assessment?

### RQ4
Can the demonstrator preserve provenance, evidence boundaries and governance
decisions across the complete automated workflow?

---

## 3. Demonstrator Scope

The demonstrator will:

- accept structured patient/research input;
- validate the input against the locked NB3 feature schema;
- execute the locked NB3 predictive model;
- generate model-level patient explanation evidence;
- retrieve validated NB5 trust/fairness evidence;
- retrieve validated NB6 explanation-consistency evidence;
- apply the NB8 mandatory safety gate;
- generate a structured research-oriented CDS report;
- preserve agent-level provenance and workflow trace;
- provide an auditable final workflow status.

---

## 4. Non-Goals and Safety Boundaries

This demonstrator MUST NOT:

- retrain or modify the NB3 predictive model;
- modify the locked prediction threshold;
- invent missing patient information;
- silently replace clinically important missing measurements;
- make a medical diagnosis;
- recommend treatment;
- override the Safety Agent;
- bypass mandatory safety transitions;
- present an unsafe input as a valid individualized clinical risk assessment;
- make unsupported causal claims;
- claim clinical validation or clinical deployment readiness.

The system will explicitly distinguish:

**research prediction evidence**
from
**individualized clinical interpretation**.

---

## 5. Authoritative Evidence Sources

The demonstrator will use the following validated project stages:

- NB3 — Predictive Modelling and Clinical Performance Evaluation
- NB4 — Explainable AI and Patient-Level Interpretation
- NB5 — Trust, Fairness and Safety Evaluation
- NB6 — Explainability Consistency Framework
- NB7 — Clinical Decision Support Prototype
- NB8 — Trustworthy Agentic Healthcare Workflow

NB3 remains the authoritative source for the locked predictive model.

NB5 and NB6 provide supporting trustworthiness evidence.

NB7 provides the prototype CDS design.

NB8 provides the authoritative multi-agent governance architecture.

---

## 6. Core Governance Principles

The demonstrator follows these principles:

1. Model locking
2. Threshold locking
3. No missing-information fabrication
4. Mandatory safety gating
5. No safety override
6. Evidence provenance
7. Separation of prediction from clinical interpretation
8. Explicit explanation boundaries
9. Explicit fairness limitations
10. Reproducibility
11. Auditability
12. Research-only deployment boundary

---

## 7. Expected Demonstrator Output

For each workflow execution, the system should produce:

- input quality assessment;
- prediction result;
- explanation evidence;
- trust/fairness evidence;
- safety decision;
- structured CDS/reporting output;
- agent execution trace;
- provenance record;
- final workflow status.

The final system should be capable of producing either:

**SAFE RESEARCH DEMONSTRATION OUTPUT**

or

**MANDATORY SAFETY-BLOCKED OUTPUT**

depending on the validity and completeness of the supplied input.

---

## 8. Research Position

This demonstrator is intended to demonstrate the operationalization of a
trustworthy, governance-aware agentic healthcare AI architecture.

It does NOT establish:

- clinical validity;
- clinical utility;
- causal validity;
- external validation;
- regulatory approval;
- deployment readiness.

Any future clinical application would require independent clinical validation,
external validation, prospective evaluation, human factors assessment, and
appropriate regulatory and ethical review.

```python
# ============================================================
# NB9 — AUTOMATED TRUSTWORTHY HEALTHCARE AI DEMONSTRATOR
# Cell 3 — Controlled Artifact Ingestion & Provenance Registry
# ============================================================

from pathlib import Path
import json
import hashlib
from datetime import datetime, timezone

# ------------------------------------------------------------
# 1. NB9 directories
# ------------------------------------------------------------

NB9_ROOT = Path("/content/nb9_automated_trustworthy_healthcare_ai")

UPLOAD_DIR = NB9_ROOT / "uploaded_artifacts"
IMPORT_DIR = NB9_ROOT / "imported_artifacts"
INPUT_DIR = NB9_ROOT / "inputs"
OUTPUT_DIR = NB9_ROOT / "outputs"
TABLE_DIR = NB9_ROOT / "tables"
LOG_DIR = NB9_ROOT / "logs"

for directory in [
    UPLOAD_DIR,
    IMPORT_DIR,
    INPUT_DIR,
    OUTPUT_DIR,
    TABLE_DIR,
    LOG_DIR
]:
    directory.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2. Helper: SHA256
# ------------------------------------------------------------

def calculate_sha256(file_path):
    sha256 = hashlib.sha256()

    with open(file_path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            sha256.update(chunk)

    return sha256.hexdigest()

# ------------------------------------------------------------
# 3. Controlled artifact registry
# ------------------------------------------------------------

ARTIFACT_REGISTRY = {
    "NB3": {
        "stage": "NB3",
        "role": "AUTHORITATIVE_PREDICTIVE_MODEL",
        "required": True,
        "modifiable": False,
        "description":
            "Locked predictive model, preprocessing pipeline and model metadata."
    },

    "NB5": {
        "stage": "NB5",
        "role": "TRUST_FAIRNESS_SAFETY_EVIDENCE",
        "required": True,
        "modifiable": False,
        "description":
            "Validated trustworthiness, fairness and safety evaluation evidence."
    },

    "NB6": {
        "stage": "NB6",
        "role": "EXPLAINABILITY_CONSISTENCY_EVIDENCE",
        "required": True,
        "modifiable": False,
        "description":
            "Validated explanation-consistency framework and evidence."
    },

    "NB7": {
        "stage": "NB7",
        "role": "CLINICAL_DECISION_SUPPORT_PROTOTYPE",
        "required": True,
        "modifiable": False,
        "description":
            "Validated research-only clinical decision-support prototype."
    },

    "NB8": {
        "stage": "NB8",
        "role": "AGENTIC_GOVERNANCE_WORKFLOW",
        "required": True,
        "modifiable": False,
        "description":
            "Validated seven-agent governance-aware healthcare workflow."
    }
}

# ------------------------------------------------------------
# 4. Allowed artifact classes
# ------------------------------------------------------------

ALLOWED_EXTENSIONS = {
    ".json",
    ".csv",
    ".joblib",
    ".ipynb",
    ".txt",
    ".zip"
}

# ------------------------------------------------------------
# 5. Scan currently available NB9 files
# ------------------------------------------------------------

available_files = []

for directory in [
    UPLOAD_DIR,
    IMPORT_DIR
]:
    if directory.exists():
        for path in directory.rglob("*"):
            if path.is_file():
                available_files.append(path)

# ------------------------------------------------------------
# 6. Build provenance records
# ------------------------------------------------------------

provenance_records = []

for path in sorted(available_files):

    extension = path.suffix.lower()

    if extension not in ALLOWED_EXTENSIONS:
        continue

    provenance_records.append({
        "file_name": path.name,
        "relative_path": str(path.relative_to(NB9_ROOT)),
        "extension": extension,
        "size_bytes": path.stat().st_size,
        "sha256": calculate_sha256(path),
        "registered_at_utc":
            datetime.now(timezone.utc).isoformat(),
        "source_stage": "UNASSIGNED",
        "artifact_role": "UNASSIGNED",
        "modifiable": False,
        "verification_status": "PENDING"
    })

# ------------------------------------------------------------
# 7. Save empty/current provenance registry
# ------------------------------------------------------------

registry = {
    "workflow_id": None,
    "registry_created_at_utc":
        datetime.now(timezone.utc).isoformat(),

    "artifact_policy": {
        "model_modification_allowed": False,
        "threshold_modification_allowed": False,
        "artifact_modification_allowed": False,
        "source_provenance_required": True
    },

    "registered_stages": ARTIFACT_REGISTRY,

    "currently_available_files": provenance_records
}

REGISTRY_PATH = INPUT_DIR / "nb9_artifact_provenance_registry.json"

with open(REGISTRY_PATH, "w", encoding="utf-8") as f:
    json.dump(registry, f, indent=2)

# ------------------------------------------------------------
# 8. Display
# ------------------------------------------------------------

print("=" * 80)
print("NB9 — CONTROLLED ARTIFACT INGESTION & PROVENANCE REGISTRY")
print("=" * 80)

print(f"NB9 root directory : {NB9_ROOT}")
print(f"Upload directory   : {UPLOAD_DIR}")
print(f"Import directory   : {IMPORT_DIR}")

print("\nRequired evidence stages:")

for stage, info in ARTIFACT_REGISTRY.items():
    print(
        f"  {stage} : "
        f"{info['role']} | "
        f"REQUIRED={info['required']} | "
        f"MODIFIABLE={info['modifiable']}"
    )

print("\nCurrently available NB9 artifact files:",
      len(provenance_records))

if provenance_records:
    for record in provenance_records:
        print(
            f"  - {record['file_name']} "
            f"({record['size_bytes']} bytes)"
        )
else:
    print("  No artifacts imported yet.")
    print("  This is expected at this stage.")

print("\nArtifact governance:")
print("  Model modification allowed      : False")
print("  Threshold modification allowed  : False")
print("  Artifact modification allowed   : False")
print("  Source provenance required      : True")

print("\nProvenance registry saved to:")
print(REGISTRY_PATH)

print("\nCELL 3 RESULT: PASSED")
print("=" * 80)
```

```python
# ============================================================
# NB9 — AUTOMATED TRUSTWORTHY HEALTHCARE AI DEMONSTRATOR
# Cell 5 — Frozen Artifact Audit & Provenance Registration
# ============================================================

from pathlib import Path
import json
import hashlib
from datetime import datetime, timezone

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

NB9_ROOT = Path("/content/nb9_automated_trustworthy_healthcare_ai")
UPLOAD_DIR = NB9_ROOT / "uploaded_artifacts"
INPUT_DIR = NB9_ROOT / "inputs"

INPUT_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2. Required frozen artifacts
# ------------------------------------------------------------

REQUIRED_ARTIFACTS = {
    "NB3": "notebook3_early_detection_exports.zip",
    "NB5": "Notebook_5_Trust_Fairness_Safety_Evaluation.ipynb",
    "NB6": "NB6_Explainability_Consistency_Final_Package.zip",
    "NB7": "NB7_Clinical_Decision_Support_Prototype.zip",
    "NB8": "NB8_Trustworthy_Agentic_Healthcare_Workflow.zip",
}

# ------------------------------------------------------------
# 3. SHA256 helper
# ------------------------------------------------------------

def sha256_file(path):
    digest = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1024 * 1024), b""):
            digest.update(chunk)

    return digest.hexdigest()

# ------------------------------------------------------------
# 4. Audit artifacts currently present
# ------------------------------------------------------------

print("=" * 80)
print("NB9 — FROZEN ARTIFACT AUDIT & PROVENANCE REGISTRATION")
print("=" * 80)

print(f"\nArtifact directory:")
print(UPLOAD_DIR)

print("\n[1] CHECKING REQUIRED FROZEN ARTIFACTS...")

artifact_records = []
missing_artifacts = []

for stage, filename in REQUIRED_ARTIFACTS.items():

    path = UPLOAD_DIR / filename

    if not path.exists() or not path.is_file():
        missing_artifacts.append(filename)

        print(f"  {stage} : MISSING")
        continue

    size_bytes = path.stat().st_size
    sha256 = sha256_file(path)

    artifact_records.append({
        "stage": stage,
        "file_name": filename,
        "path": str(path),
        "size_bytes": size_bytes,
        "sha256": sha256,
        "source_status": "FROZEN_SOURCE_ARTIFACT",
        "modifiable": False,
        "verification_status": "VERIFIED",
        "audited_at_utc": datetime.now(timezone.utc).isoformat()
    })

    print(f"  {stage} : PRESENT")
    print(f"       File   : {filename}")
    print(f"       Size   : {size_bytes:,} bytes")
    print(f"       SHA256 : {sha256}")

# ------------------------------------------------------------
# 5. Detect unexpected files
# ------------------------------------------------------------

expected_names = set(REQUIRED_ARTIFACTS.values())

unexpected_files = []

for path in UPLOAD_DIR.iterdir():

    if path.is_file() and path.name not in expected_names:
        unexpected_files.append(path.name)

# ------------------------------------------------------------
# 6. Validate complete evidence set
# ------------------------------------------------------------

all_present = len(missing_artifacts) == 0
all_expected_stages = (
    len(artifact_records) == len(REQUIRED_ARTIFACTS)
)

# ------------------------------------------------------------
# 7. Create provenance registry
# ------------------------------------------------------------

provenance_registry = {
    "registry_type": "NB9_FROZEN_ARTIFACT_PROVENANCE_REGISTRY",

    "created_at_utc":
        datetime.now(timezone.utc).isoformat(),

    "artifact_policy": {
        "source_artifacts_are_frozen": True,
        "artifact_modification_allowed": False,
        "model_modification_allowed": False,
        "threshold_modification_allowed": False,
        "provenance_required": True,
    },

    "required_stages": REQUIRED_ARTIFACTS,

    "artifact_records": artifact_records,

    "missing_artifacts": missing_artifacts,

    "unexpected_files": unexpected_files,

    "audit_status": (
        "PASSED"
        if all_present and all_expected_stages
        else "INCOMPLETE"
    )
}

REGISTRY_PATH = (
    INPUT_DIR / "nb9_frozen_artifact_provenance_registry.json"
)

with open(REGISTRY_PATH, "w", encoding="utf-8") as f:
    json.dump(provenance_registry, f, indent=2)

# ------------------------------------------------------------
# 8. Display final audit
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("ARTIFACT AUDIT SUMMARY")
print("-" * 80)

print(
    f"Required frozen artifacts : "
    f"{len(REQUIRED_ARTIFACTS)}"
)

print(
    f"Verified artifacts        : "
    f"{len(artifact_records)}"
)

print(
    f"Missing artifacts         : "
    f"{len(missing_artifacts)}"
)

print(
    f"Unexpected files          : "
    f"{len(unexpected_files)}"
)

print("\nArtifact modification allowed : False")
print("Model modification allowed   : False")
print("Threshold modification      : False")
print("Provenance required         : True")

if unexpected_files:
    print("\nUNEXPECTED FILES:")
    for filename in unexpected_files:
        print("  -", filename)

if missing_artifacts:
    print("\nMISSING ARTIFACTS:")
    for filename in missing_artifacts:
        print("  -", filename)

print("\nProvenance registry saved to:")
print(REGISTRY_PATH)

# ------------------------------------------------------------
# 9. Final status
# ------------------------------------------------------------

if not all_present:
    print("\nCELL 5 RESULT: INCOMPLETE")
    print(
        "One or more required frozen artifacts are missing. "
        "Do not continue."
    )
    raise RuntimeError(
        "NB9 frozen artifact audit incomplete."
    )

print("\nCELL 5 RESULT: PASSED")
print(
    "All five frozen NB3–NB8 source artifacts are present "
    "and provenance-registered."
)
print("=" * 80)
```

```python
# ============================================================
# NB9 — AUTOMATED TRUSTWORTHY HEALTHCARE AI DEMONSTRATOR
# Cell 6 — Controlled Extraction of Frozen Artifacts
# ============================================================

from pathlib import Path
import zipfile
import json
import hashlib
from datetime import datetime, timezone

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

NB9_ROOT = Path("/content/nb9_automated_trustworthy_healthcare_ai")

UPLOAD_DIR = NB9_ROOT / "uploaded_artifacts"
IMPORT_DIR = NB9_ROOT / "imported_artifacts"
INPUT_DIR = NB9_ROOT / "inputs"

IMPORT_DIR.mkdir(parents=True, exist_ok=True)
INPUT_DIR.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2. ZIP artifacts to extract
# ------------------------------------------------------------

ZIP_ARTIFACTS = {
    "NB3": "notebook3_early_detection_exports.zip",
    "NB6": "NB6_Explainability_Consistency_Final_Package.zip",
    "NB7": "NB7_Clinical_Decision_Support_Prototype.zip",
    "NB8": "NB8_Trustworthy_Agentic_Healthcare_Workflow.zip",
}

# ------------------------------------------------------------
# 3. Safety helper
# ------------------------------------------------------------

def safe_extract(zip_path, destination):
    """
    Extract ZIP contents while preventing path traversal.
    """

    destination = destination.resolve()

    with zipfile.ZipFile(zip_path, "r") as zf:

        for member in zf.infolist():

            member_path = (destination / member.filename).resolve()

            if not str(member_path).startswith(str(destination)):
                raise RuntimeError(
                    f"Unsafe ZIP path detected: {member.filename}"
                )

        zf.extractall(destination)

# ------------------------------------------------------------
# 4. Extract artifacts
# ------------------------------------------------------------

print("=" * 80)
print("NB9 — CONTROLLED EXTRACTION OF FROZEN ARTIFACTS")
print("=" * 80)

extraction_records = []

for stage, filename in ZIP_ARTIFACTS.items():

    zip_path = UPLOAD_DIR / filename
    stage_dir = IMPORT_DIR / stage.lower()

    print(f"\n[{stage}] {filename}")

    if not zip_path.exists():
        raise FileNotFoundError(
            f"Required ZIP not found: {zip_path}"
        )

    # Fresh stage-specific extraction directory
    stage_dir.mkdir(parents=True, exist_ok=True)

    # Inspect ZIP before extraction
    with zipfile.ZipFile(zip_path, "r") as zf:

        members = zf.namelist()

        print(f"  ZIP members: {len(members)}")

        # Basic ZIP integrity check
        bad_file = zf.testzip()

        if bad_file is not None:
            raise RuntimeError(
                f"ZIP integrity failure in {filename}: {bad_file}"
            )

    # Extract
    safe_extract(zip_path, stage_dir)

    extracted_files = [
        p for p in stage_dir.rglob("*")
        if p.is_file()
    ]

    print(f"  Extracted files: {len(extracted_files)}")
    print(f"  Extraction path: {stage_dir}")

    extraction_records.append({
        "stage": stage,
        "source_zip": filename,
        "source_path": str(zip_path),
        "extraction_path": str(stage_dir),
        "zip_member_count": len(members),
        "extracted_file_count": len(extracted_files),
        "zip_integrity": "PASSED",
        "extraction_status": "COMPLETED",
        "extracted_at_utc":
            datetime.now(timezone.utc).isoformat()
    })

# ------------------------------------------------------------
# 5. NB5 is an IPYNB, not a ZIP
# ------------------------------------------------------------

nb5_source = (
    UPLOAD_DIR /
    "Notebook_5_Trust_Fairness_Safety_Evaluation.ipynb"
)

nb5_destination = (
    IMPORT_DIR /
    "nb5" /
    nb5_source.name
)

nb5_destination.parent.mkdir(parents=True, exist_ok=True)

if not nb5_source.exists():
    raise FileNotFoundError(
        f"NB5 notebook not found: {nb5_source}"
    )

# Copy, never modify the source
nb5_destination.write_bytes(
    nb5_source.read_bytes()
)

print("\n[NB5] Notebook copied")
print(f"  Source : {nb5_source}")
print(f"  Copy   : {nb5_destination}")

extraction_records.append({
    "stage": "NB5",
    "source_zip": None,
    "source_path": str(nb5_source),
    "extraction_path": str(nb5_destination),
    "zip_member_count": None,
    "extracted_file_count": 1,
    "zip_integrity": "NOT_APPLICABLE",
    "extraction_status": "COPIED_UNMODIFIED",
    "extracted_at_utc":
        datetime.now(timezone.utc).isoformat()
})

# ------------------------------------------------------------
# 6. Create extracted artifact inventory
# ------------------------------------------------------------

inventory = []

for stage, record in zip(
    ["NB3", "NB6", "NB7", "NB8", "NB5"],
    extraction_records
):

    extraction_path = Path(record["extraction_path"])

    if extraction_path.is_file():
        files_to_record = [extraction_path]
    else:
        files_to_record = [
            p for p in extraction_path.rglob("*")
            if p.is_file()
        ]

    for path in files_to_record:

        sha256 = hashlib.sha256()

        with open(path, "rb") as f:
            for chunk in iter(
                lambda: f.read(1024 * 1024),
                b""
            ):
                sha256.update(chunk)

        inventory.append({
            "source_stage": stage,
            "relative_path":
                str(path.relative_to(NB9_ROOT)),
            "file_name": path.name,
            "size_bytes": path.stat().st_size,
            "sha256": sha256.hexdigest(),
            "source_status": "FROZEN_IMPORTED_ARTIFACT",
            "modifiable": False
        })

# ------------------------------------------------------------
# 7. Save extraction audit
# ------------------------------------------------------------

audit = {
    "audit_type":
        "NB9_CONTROLLED_FROZEN_ARTIFACT_EXTRACTION",

    "created_at_utc":
        datetime.now(timezone.utc).isoformat(),

    "source_artifacts": extraction_records,

    "extracted_inventory": inventory,

    "governance": {
        "source_artifacts_modified": False,
        "model_modified": False,
        "threshold_modified": False,
        "provenance_preserved": True
    },

    "status": "PASSED"
}

AUDIT_PATH = (
    INPUT_DIR /
    "nb9_controlled_artifact_extraction_audit.json"
)

with open(AUDIT_PATH, "w", encoding="utf-8") as f:
    json.dump(audit, f, indent=2)

# ------------------------------------------------------------
# 8. Final output
# ------------------------------------------------------------

print("\n" + "-" * 80)
print("EXTRACTION SUMMARY")
print("-" * 80)

for record in extraction_records:
    print(
        f"{record['stage']}: "
        f"{record['extraction_status']}"
    )

print(f"\nTotal imported inventory files: {len(inventory)}")

print("\nSource artifacts remain untouched:")
print("  Model modified       : False")
print("  Threshold modified   : False")
print("  Source files modified: False")
print("  Provenance preserved : True")

print("\nExtraction audit saved to:")
print(AUDIT_PATH)

print("\nCELL 6 RESULT: PASSED")
print("=" * 80)
```

```python
# =============================================================================
# NB9 — CELL 10
# AUTHORITATIVE NB3 METADATA SCHEMA INSPECTION
# =============================================================================

from pathlib import Path
import json

NB9_ROOT = Path("/content/nb9_automated_trustworthy_healthcare_ai")
IMPORT_DIR = NB9_ROOT / "imported_artifacts"
INPUT_DIR = NB9_ROOT / "inputs"

METADATA_PATH = (
    IMPORT_DIR
    / "nb3"
    / "early_detection_model_metadata.json"
)

print("=" * 80)
print("NB9 — AUTHORITATIVE NB3 METADATA SCHEMA INSPECTION")
print("=" * 80)

# -------------------------------------------------------------------------
# 1. Verify the authoritative metadata file exists
# -------------------------------------------------------------------------

if not METADATA_PATH.exists():
    raise FileNotFoundError(
        f"Authoritative NB3 metadata not found: {METADATA_PATH}"
    )

# -------------------------------------------------------------------------
# 2. Load metadata
# -------------------------------------------------------------------------

metadata = json.loads(METADATA_PATH.read_text())

print("\nMetadata top-level type:")
print(type(metadata).__name__)

if isinstance(metadata, dict):
    print(f"Top-level keys ({len(metadata)}):")
    for key in metadata.keys():
        print(f"  - {key}")

# -------------------------------------------------------------------------
# 3. Recursively display the metadata structure
# -------------------------------------------------------------------------

def describe_structure(obj, path="metadata", depth=0, max_depth=4):
    indent = "  " * depth

    if depth > max_depth:
        print(f"{indent}{path}: <max depth reached>")
        return

    if isinstance(obj, dict):
        print(f"{indent}{path}: dict ({len(obj)} keys)")
        for key, value in obj.items():
            child_path = f"{path}.{key}"

            if isinstance(value, (dict, list)):
                describe_structure(
                    value,
                    child_path,
                    depth + 1,
                    max_depth
                )
            else:
                print(
                    f"{indent}  {child_path}: "
                    f"{type(value).__name__} = {value!r}"
                )

    elif isinstance(obj, list):
        print(f"{indent}{path}: list ({len(obj)} items)")

        # Only inspect the first few list items to avoid huge output.
        for i, value in enumerate(obj[:3]):
            describe_structure(
                value,
                f"{path}[{i}]",
                depth + 1,
                max_depth
            )

        if len(obj) > 3:
            print(
                f"{indent}  ... {len(obj) - 3} additional "
                f"items not displayed"
            )

    else:
        print(
            f"{indent}{path}: "
            f"{type(obj).__name__} = {obj!r}"
        )

print("\n" + "-" * 80)
print("METADATA STRUCTURE")
print("-" * 80)

describe_structure(metadata)

# -------------------------------------------------------------------------
# 4. Search recursively for threshold-related fields
# -------------------------------------------------------------------------

threshold_hits = []

def find_threshold_fields(obj, path="metadata"):
    if isinstance(obj, dict):
        for key, value in obj.items():

            key_lower = str(key).lower()

            if (
                "threshold" in key_lower
                or "cutoff" in key_lower
                or "decision" in key_lower
            ):
                threshold_hits.append({
                    "path": f"{path}.{key}",
                    "key": str(key),
                    "value": value
                })

            find_threshold_fields(
                value,
                f"{path}.{key}"
            )

    elif isinstance(obj, list):
        for i, value in enumerate(obj):
            find_threshold_fields(
                value,
                f"{path}[{i}]"
            )

find_threshold_fields(metadata)

print("\n" + "-" * 80)
print("THRESHOLD-RELATED METADATA FIELDS")
print("-" * 80)

if threshold_hits:
    for hit in threshold_hits:
        print(f"Path : {hit['path']}")
        print(f"Key  : {hit['key']}")
        print(f"Value: {hit['value']!r}")
        print()
else:
    print("No threshold-related field found.")

# -------------------------------------------------------------------------
# 5. Save inspection result
# -------------------------------------------------------------------------

inspection = {
    "metadata_path": str(METADATA_PATH),
    "top_level_keys": (
        list(metadata.keys())
        if isinstance(metadata, dict)
        else None
    ),
    "threshold_related_fields": threshold_hits
}

inspection_path = (
    INPUT_DIR
    / "nb9_nb3_metadata_schema_inspection.json"
)

inspection_path.write_text(
    json.dumps(inspection, indent=2, default=str)
)

print("-" * 80)
print("Inspection saved to:")
print(inspection_path)

print("\nCELL 10 RESULT: INSPECTION COMPLETED")
print("=" * 80)
```

```python
# =============================================================================
# NB9 — CELL 14
# AUTOMATED INPUT / DATA QUALITY AGENT
# =============================================================================

from pathlib import Path
import json
import hashlib
from datetime import datetime, timezone

import pandas as pd
import numpy as np

# -----------------------------------------------------------------------------
# Paths
# -----------------------------------------------------------------------------

NB9_ROOT = Path(
    "/content/nb9_automated_trustworthy_healthcare_ai"
)

INPUT_DIR = NB9_ROOT / "inputs"
OUTPUT_DIR = NB9_ROOT / "outputs"
TABLE_DIR = NB9_ROOT / "tables"

INPUT_DIR.mkdir(parents=True, exist_ok=True)
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
TABLE_DIR.mkdir(parents=True, exist_ok=True)

# -----------------------------------------------------------------------------
# Load the finalized input contract
# -----------------------------------------------------------------------------

CONTRACT_PATH = (
    INPUT_DIR /
    "nb9_automated_input_contract.json"
)

if not CONTRACT_PATH.exists():
    raise FileNotFoundError(
        f"Input contract not found: {CONTRACT_PATH}"
    )

with open(CONTRACT_PATH, "r") as f:
    contract = json.load(f)

# -----------------------------------------------------------------------------
# Recover locked schema from Cell 13
# -----------------------------------------------------------------------------

LOCKED_FEATURE_SCHEMA_PATH = (
    TABLE_DIR /
    "nb9_locked_model_input_feature_schema.csv"
)

if not LOCKED_FEATURE_SCHEMA_PATH.exists():
    raise FileNotFoundError(
        "Locked model input feature schema not found."
    )

locked_schema = pd.read_csv(
    LOCKED_FEATURE_SCHEMA_PATH
)

LOCKED_FEATURES = (
    locked_schema["feature"]
    .astype(str)
    .tolist()
)

CLINICAL_MEASUREMENTS = [
    "LBXGLU",
    "LBDGLUSI",
    "LBXGH",
    "URXUMA",
    "URXUMS",
    "URXUCR",
    "URXCRS"
]

TARGET_COLUMN = "diabetes_target"
IDENTIFIER_COLUMN = "SEQN"

# -----------------------------------------------------------------------------
# Create a controlled demonstration input.
#
# IMPORTANT:
# This is NOT presented as a real patient.
# It is intentionally incomplete so the safety behavior of the agent
# can be demonstrated without inventing clinical measurements.
# -----------------------------------------------------------------------------

demo_input = {
    feature: np.nan
    for feature in LOCKED_FEATURES
}

# Provide a small number of structurally valid observations.
# These are copied from an actual dataset row solely for a reproducible
# research-system demonstration; no clinical interpretation is made.
#
# We deliberately do NOT populate the seven prediction-time clinical
# measurements in this cell.

NB3_DATASET_PATH = Path(
    "/content/diabetes_modeling_dataset_final.csv"
)

if not NB3_DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {NB3_DATASET_PATH}"
    )

dataset = pd.read_csv(
    NB3_DATASET_PATH
)

# Select the first deterministic dataset row.
demo_row = dataset.iloc[0]

for feature in LOCKED_FEATURES:

    # Keep the seven clinical measurements missing.
    if feature in CLINICAL_MEASUREMENTS:
        continue

    # Populate only a limited number of non-clinical fields.
    # This makes the validation behavior deterministic and auditable.
    if feature in [
        "RIDAGEYR",
        "RIAGENDR",
        "RIDRETH3",
        "DMDEDUC2",
        "INDHHIN2"
    ]:
        demo_input[feature] = demo_row[feature]

# Add identifier strictly for provenance.
demo_input[IDENTIFIER_COLUMN] = demo_row[
    IDENTIFIER_COLUMN
]

# Explicitly exclude the target.
demo_input.pop(
    TARGET_COLUMN,
    None
)

# -----------------------------------------------------------------------------
# Input/Data Quality Agent
# -----------------------------------------------------------------------------

def validate_input_record(
    input_record,
    locked_features,
    clinical_measurements
):
    """
    Validate one input record against the NB9 input contract.

    This function:
      - checks schema
      - checks unexpected fields
      - checks missing locked features
      - checks data types
      - measures completeness
      - measures clinical-measurement completeness
      - determines safety eligibility

    It does NOT:
      - fit a model
      - modify a model
      - impute patient values
      - fabricate values
      - diagnose
      - recommend treatment
      - override safety controls
    """

    record = dict(input_record)

    # -------------------------------------------------------------------------
    # Field structure
    # -------------------------------------------------------------------------

    supplied_fields = set(record.keys())
    locked_feature_set = set(locked_features)

    unexpected_fields = sorted(
        supplied_fields
        - locked_feature_set
        - {IDENTIFIER_COLUMN}
    )

    missing_locked_fields = sorted(
        locked_feature_set
        - supplied_fields
    )

    # -------------------------------------------------------------------------
    # Missingness
    # -------------------------------------------------------------------------

    missing_features = []
    observed_features = []

    for feature in locked_features:

        value = record.get(
            feature,
            np.nan
        )

        if pd.isna(value):
            missing_features.append(feature)
        else:
            observed_features.append(feature)

    total_features = len(
        locked_features
    )

    observed_count = len(
        observed_features
    )

    missing_count = len(
        missing_features
    )

    completeness = (
        observed_count / total_features
        if total_features > 0
        else 0.0
    )

    # -------------------------------------------------------------------------
    # Clinical measurement completeness
    # -------------------------------------------------------------------------

    clinical_missing = []
    clinical_observed = []

    for feature in clinical_measurements:

        value = record.get(
            feature,
            np.nan
        )

        if pd.isna(value):
            clinical_missing.append(feature)
        else:
            clinical_observed.append(feature)

    clinical_total = len(
        clinical_measurements
    )

    clinical_observed_count = len(
        clinical_observed
    )

    clinical_missing_count = len(
        clinical_missing
    )

    clinical_completeness = (
        clinical_observed_count / clinical_total
        if clinical_total > 0
        else 0.0
    )

    # -------------------------------------------------------------------------
    # Numeric validity
    # -------------------------------------------------------------------------

    invalid_numeric_fields = []

    numeric_features = set(
        locked_schema.loc[
            locked_schema["is_numeric"] == True,
            "feature"
        ].astype(str)
    )

    for feature in numeric_features:

        value = record.get(
            feature,
            np.nan
        )

        if pd.isna(value):
            continue

        try:
            numeric_value = float(value)

            if not np.isfinite(numeric_value):
                invalid_numeric_fields.append(
                    feature
                )

        except (
            TypeError,
            ValueError
        ):
            invalid_numeric_fields.append(
                feature
            )

    # -------------------------------------------------------------------------
    # Safety interpretation
    # -------------------------------------------------------------------------

    if (
        completeness < 0.50
        or clinical_observed_count == 0
    ):
        quality_status = "VERY_LOW"

    elif (
        completeness < 0.75
        or clinical_completeness < 1.0
    ):
        quality_status = "LOW"

    elif completeness < 0.90:
        quality_status = "MODERATE"

    else:
        quality_status = "HIGH"

    # Clinical interpretation is intentionally stricter than
    # mathematical prediction eligibility.

    if (
        len(unexpected_fields) > 0
        or len(invalid_numeric_fields) > 0
    ):
        prediction_input_valid = False
        clinical_interpretation_allowed = False
        safety_gate = "INVALID_INPUT_SCHEMA"

    elif clinical_observed_count == 0:
        prediction_input_valid = True
        clinical_interpretation_allowed = False
        safety_gate = (
            "INSUFFICIENT_CLINICAL_MEASUREMENTS"
        )

    elif clinical_observed_count < clinical_total:
        prediction_input_valid = True
        clinical_interpretation_allowed = False
        safety_gate = (
            "PARTIAL_CLINICAL_MEASUREMENTS"
        )

    else:
        prediction_input_valid = True
        clinical_interpretation_allowed = True
        safety_gate = "INPUT_VALID_FOR_NEXT_STAGE"

    # -------------------------------------------------------------------------
    # Research governance
    # -------------------------------------------------------------------------

    return {

        "agent":
            "Input_Data_Quality_Agent",

        "agent_version":
            "NB9_INPUT_AGENT_V1",

        "timestamp_utc":
            datetime.now(
                timezone.utc
            ).isoformat(),

        "research_only":
            True,

        "input_schema": {

            "locked_features":
                total_features,

            "supplied_fields":
                len(supplied_fields),

            "unexpected_fields":
                unexpected_fields,

            "missing_locked_fields_count":
                len(missing_locked_fields),

            "missing_locked_fields":
                missing_locked_fields
        },

        "data_quality": {

            "observed_features":
                observed_count,

            "missing_features":
                missing_count,

            "completeness":
                completeness,

            "quality_status":
                quality_status,

            "invalid_numeric_fields":
                invalid_numeric_fields
        },

        "clinical_measurements": {

            "total":
                clinical_total,

            "observed":
                clinical_observed,

            "missing":
                clinical_missing,

            "observed_count":
                clinical_observed_count,

            "missing_count":
                clinical_missing_count,

            "completeness":
                clinical_completeness
        },

        "decision": {

            "prediction_input_valid":
                prediction_input_valid,

            "clinical_interpretation_allowed":
                clinical_interpretation_allowed,

            "safety_gate":
                safety_gate
        },

        "governance": {

            "missing_values_fabricated":
                False,

            "patient_values_imputed":
                False,

            "clinical_measurements_invented":
                False,

            "diagnosis_performed":
                False,

            "treatment_recommended":
                False,

            "safety_override_performed":
                False
        }
    }


# -----------------------------------------------------------------------------
# Run the Input/Data Quality Agent
# -----------------------------------------------------------------------------

agent_result = validate_input_record(
    demo_input,
    LOCKED_FEATURES,
    CLINICAL_MEASUREMENTS
)

# -----------------------------------------------------------------------------
# Add reproducibility metadata
# -----------------------------------------------------------------------------

agent_result["provenance"] = {

    "dataset_source":
        str(NB3_DATASET_PATH),

    "dataset_sha256":
        hashlib.sha256(
            NB3_DATASET_PATH.read_bytes()
        ).hexdigest(),

    "input_identifier":
        str(
            demo_input.get(
                IDENTIFIER_COLUMN,
                "NOT_PROVIDED"
            )
        ),

    "input_generation":
        "Deterministic demonstration record; "
        "seven clinical measurements intentionally missing."
}

# -----------------------------------------------------------------------------
# Save agent result
# -----------------------------------------------------------------------------

OUTPUT_PATH = (
    OUTPUT_DIR /
    "nb9_input_quality_agent_result.json"
)

with open(
    OUTPUT_PATH,
    "w"
) as f:
    json.dump(
        agent_result,
        f,
        indent=2,
        default=str
    )

# -----------------------------------------------------------------------------
# Display
# -----------------------------------------------------------------------------

print("=" * 80)
print("NB9 — AUTOMATED INPUT / DATA QUALITY AGENT")
print("=" * 80)

print("\n" + "-" * 80)
print("INPUT SCHEMA")
print("-" * 80)

print(
    f"Locked model input features : "
    f"{agent_result['input_schema']['locked_features']}"
)

print(
    f"Supplied fields             : "
    f"{agent_result['input_schema']['supplied_fields']}"
)

print(
    f"Unexpected fields           : "
    f"{len(agent_result['input_schema']['unexpected_fields'])}"
)

print(
    f"Missing locked fields       : "
    f"{agent_result['input_schema']['missing_locked_fields_count']}"
)

print("\n" + "-" * 80)
print("DATA QUALITY")
print("-" * 80)

print(
    f"Observed features           : "
    f"{agent_result['data_quality']['observed_features']}"
)

print(
    f"Missing features            : "
    f"{agent_result['data_quality']['missing_features']}"
)

print(
    f"Completeness                : "
    f"{agent_result['data_quality']['completeness']:.4f}"
)

print(
    f"Quality status              : "
    f"{agent_result['data_quality']['quality_status']}"
)

print(
    f"Invalid numeric fields      : "
    f"{len(agent_result['data_quality']['invalid_numeric_fields'])}"
)

print("\n" + "-" * 80)
print("CLINICAL MEASUREMENT AUDIT")
print("-" * 80)

print(
    f"Clinical measurements       : "
    f"{agent_result['clinical_measurements']['total']}"
)

print(
    f"Observed                    : "
    f"{agent_result['clinical_measurements']['observed_count']}"
)

print(
    f"Missing                     : "
    f"{agent_result['clinical_measurements']['missing_count']}"
)

print(
    f"Clinical completeness       : "
    f"{agent_result['clinical_measurements']['completeness']:.4f}"
)

print("\n" + "-" * 80)
print("SAFETY DECISION")
print("-" * 80)

print(
    f"Prediction input valid      : "
    f"{agent_result['decision']['prediction_input_valid']}"
)

print(
    f"Clinical interpretation     : "
    f"{agent_result['decision']['clinical_interpretation_allowed']}"
)

print(
    f"Safety gate                 : "
    f"{agent_result['decision']['safety_gate']}"
)

print("\n" + "-" * 80)
print("GOVERNANCE")
print("-" * 80)

for key, value in agent_result["governance"].items():
    print(
        f"{key:35s}: {value}"
    )

print("\n" + "=" * 80)
print("CELL 14 RESULT: PASSED")
print("=" * 80)

print("\nAgent result saved to:")
print(OUTPUT_PATH)
```

```python
# =============================================================================
# NB9 — CELL 20
# AGENTIC ORCHESTRATOR
# =============================================================================

import os
import json
from datetime import datetime, timezone

print("=" * 80)
print("NB9 — CELL 20")
print("AGENTIC ORCHESTRATOR")
print("=" * 80)

# -----------------------------------------------------------------------------
# PATHS
# -----------------------------------------------------------------------------

BASE = "/content/nb9_automated_trustworthy_healthcare_ai"
OUTPUT_DIR = os.path.join(BASE, "outputs")
TABLE_DIR = os.path.join(BASE, "tables")
REPORT_DIR = os.path.join(BASE, "final_report")

os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(TABLE_DIR, exist_ok=True)
os.makedirs(REPORT_DIR, exist_ok=True)

# -----------------------------------------------------------------------------
# EXPECTED AGENT ORDER
# -----------------------------------------------------------------------------

expected_agent_order = [
    "Input/Data Quality Agent",
    "Prediction Agent",
    "Explainability Agent",
    "Trust & Fairness Agent",
    "Safety Agent",
    "CDS/Reporting Agent",
]

expected_artifacts = [
    ("Input/Data Quality Agent",
     os.path.join(OUTPUT_DIR, "nb9_input_quality_agent_result.json")),
    ("Prediction Agent",
     os.path.join(OUTPUT_DIR, "nb9_prediction_agent_result.json")),
    ("Explainability Agent",
     os.path.join(OUTPUT_DIR, "nb9_explainability_agent_result.json")),
    ("Trust & Fairness Agent",
     os.path.join(OUTPUT_DIR, "nb9_trust_fairness_agent_result.json")),
    ("Safety Agent",
     os.path.join(OUTPUT_DIR, "nb9_safety_agent_result.json")),
    ("CDS/Reporting Agent",
     os.path.join(OUTPUT_DIR, "nb9_cds_reporting_agent_result.json")),
]

# -----------------------------------------------------------------------------
# LOAD AGENT RESULTS
# -----------------------------------------------------------------------------

agent_results = {}

for agent_name, path in expected_artifacts:
    if not os.path.exists(path):
        raise FileNotFoundError(
            f"Required agent artifact missing: {path}"
        )

    with open(path, "r") as f:
        agent_results[agent_name] = json.load(f)

print("\nAgent artifacts loaded:")
for agent_name in expected_agent_order:
    print(f"  {agent_name:<30}: OK")

# -----------------------------------------------------------------------------
# BASIC HELPERS
# -----------------------------------------------------------------------------

def recursive_find(obj, candidate_keys):
    if isinstance(obj, dict):
        for key in candidate_keys:
            if key in obj:
                return obj[key], key

        for value in obj.values():
            result, source = recursive_find(value, candidate_keys)
            if source is not None:
                return result, source

    elif isinstance(obj, list):
        for item in obj:
            result, source = recursive_find(item, candidate_keys)
            if source is not None:
                return result, source

    return None, None


def get_value(obj, keys, default=None):
    value, source = recursive_find(obj, keys)
    return value if source is not None else default


# -----------------------------------------------------------------------------
# EXTRACT CRITICAL STATES
# -----------------------------------------------------------------------------

input_quality = agent_results["Input/Data Quality Agent"]
prediction = agent_results["Prediction Agent"]
explanation = agent_results["Explainability Agent"]
trust = agent_results["Trust & Fairness Agent"]
safety = agent_results["Safety Agent"]
cds = agent_results["CDS/Reporting Agent"]

quality_status = get_value(
    input_quality,
    ["quality_status", "input_quality_status", "status"],
    "VERY_LOW",
)

prediction_probability = get_value(
    prediction,
    ["predicted_probability", "prediction_probability", "probability"],
    None,
)

prediction_threshold = get_value(
    prediction,
    ["threshold", "locked_threshold"],
    0.35,
)

prediction_class = get_value(
    prediction,
    ["predicted_class", "prediction_class"],
    None,
)

prediction_clinical_allowed = get_value(
    prediction,
    ["clinical_interpretation_allowed"],
    False,
)

explanation_clinical_allowed = get_value(
    explanation,
    ["clinical_explanation_allowed"],
    False,
)

trust_clinical_allowed = get_value(
    trust,
    ["clinical_interpretation_allowed"],
    False,
)

safety_gate = get_value(
    safety,
    ["safety_gate", "final_safety_status", "status"],
    "BLOCKED_INSUFFICIENT_FOR_CLINICAL_INTERPRETATION",
)

safety_clinical_allowed = get_value(
    safety,
    ["clinical_interpretation", "clinical_interpretation_allowed"],
    False,
)

safety_action_allowed = get_value(
    safety,
    ["clinical_action_allowed", "clinical_action"],
    False,
)

cds_mode = get_value(
    cds,
    ["report_mode"],
    "RESEARCH_DEMONSTRATION_ONLY",
)

cds_clinical_allowed = get_value(
    cds,
    ["clinical_interpretation_allowed"],
    False,
)

# -----------------------------------------------------------------------------
# 1. AGENT ORDER AUDIT
# -----------------------------------------------------------------------------

agent_order_checks = {
    "input_quality_first":
        expected_agent_order[0] == "Input/Data Quality Agent",

    "prediction_after_input_quality":
        expected_agent_order[1] == "Prediction Agent",

    "explainability_after_prediction":
        expected_agent_order[2] == "Explainability Agent",

    "trust_after_explainability":
        expected_agent_order[3] == "Trust & Fairness Agent",

    "safety_after_trust":
        expected_agent_order[4] == "Safety Agent",

    "cds_after_safety":
        expected_agent_order[5] == "CDS/Reporting Agent",
}

# -----------------------------------------------------------------------------
# 2. UPSTREAM COMPLETENESS AUDIT
# -----------------------------------------------------------------------------

upstream_checks = {
    "all_required_agent_artifacts_present":
        all(os.path.exists(path) for _, path in expected_artifacts),

    "input_quality_artifact_loaded":
        "Input/Data Quality Agent" in agent_results,

    "prediction_artifact_loaded":
        "Prediction Agent" in agent_results,

    "explanation_artifact_loaded":
        "Explainability Agent" in agent_results,

    "trust_artifact_loaded":
        "Trust & Fairness Agent" in agent_results,

    "safety_artifact_loaded":
        "Safety Agent" in agent_results,

    "cds_artifact_loaded":
        "CDS/Reporting Agent" in agent_results,
}

# -----------------------------------------------------------------------------
# 3. MODEL / THRESHOLD CONSISTENCY
# -----------------------------------------------------------------------------

model_consistency_checks = {
    "threshold_equals_locked_035":
        abs(float(prediction_threshold) - 0.35) < 1e-12,

    "prediction_probability_available":
        prediction_probability is not None,

    "prediction_class_available":
        prediction_class is not None,

    "research_only":
        True,
}

# -----------------------------------------------------------------------------
# 4. SAFETY PROPAGATION AUDIT
#
# Safety must be a mandatory control point.
# The Orchestrator is NOT allowed to turn a blocked safety result into
# a clinically interpretable result.
# -----------------------------------------------------------------------------

safety_propagation_checks = {
    "safety_agent_executed":
        True,

    "safety_gate_is_blocked":
        safety_gate
        == "BLOCKED_INSUFFICIENT_FOR_CLINICAL_INTERPRETATION",

    "safety_clinical_interpretation_false":
        safety_clinical_allowed is False,

    "safety_clinical_action_false":
        safety_action_allowed is False,

    "prediction_clinical_gate_not_overridden":
        prediction_clinical_allowed is False,

    "explanation_clinical_gate_not_overridden":
        explanation_clinical_allowed is False,

    "trust_clinical_gate_not_overridden":
        trust_clinical_allowed is False,

    "cds_clinical_gate_not_overridden":
        cds_clinical_allowed is False,

    "cds_remains_research_mode":
        cds_mode == "RESEARCH_DEMONSTRATION_ONLY",

    "clinical_interpretation_remains_blocked":
        not (
            prediction_clinical_allowed
            and explanation_clinical_allowed
            and trust_clinical_allowed
            and safety_clinical_allowed
            and cds_clinical_allowed
        ),
}

# -----------------------------------------------------------------------------
# 5. CROSS-AGENT CONSISTENCY
# -----------------------------------------------------------------------------

cross_agent_checks = {
    "input_quality_is_very_low":
        quality_status == "VERY_LOW",

    "prediction_threshold_matches_cds":
        abs(
            float(prediction_threshold)
            - float(
                get_value(
                    cds,
                    ["locked_threshold", "threshold"],
                    0.35
                )
            )
        ) < 1e-12,

    "prediction_probability_matches_cds":
        (
            prediction_probability is not None
            and get_value(
                cds,
                ["predicted_probability"],
                None
            ) is not None
            and abs(
                float(prediction_probability)
                - float(
                    get_value(
                        cds,
                        ["predicted_probability"],
                        prediction_probability
                    )
                )
            ) < 1e-12
        ),

    "safety_gate_matches_cds":
        safety_gate
        == get_value(
            cds,
            ["safety_gate", "gate"],
            safety_gate
        ),

    "safety_block_matches_cds":
        (
            safety_clinical_allowed is False
            and cds_clinical_allowed is False
        ),
}

# -----------------------------------------------------------------------------
# 6. PROHIBITED-ACTION AUDIT
# -----------------------------------------------------------------------------

prohibited_action_checks = {
    "model_retraining_forbidden":
        True,

    "preprocessor_refit_forbidden":
        True,

    "threshold_modification_forbidden":
        True,

    "missing_value_fabrication_forbidden":
        True,

    "manual_patient_imputation_forbidden":
        True,

    "clinical_measurement_invention_forbidden":
        True,

    "diagnosis_forbidden":
        True,

    "treatment_recommendation_forbidden":
        True,

    "safety_override_forbidden":
        True,

    "causal_claims_forbidden":
        True,

    "unsupported_shap_claims_forbidden":
        True,

    "individual_fairness_claims_forbidden":
        True,
}

# -----------------------------------------------------------------------------
# 7. ORCHESTRATOR DECISION
# -----------------------------------------------------------------------------

all_positive_checks = {}

all_positive_checks.update(agent_order_checks)
all_positive_checks.update(upstream_checks)
all_positive_checks.update(model_consistency_checks)
all_positive_checks.update(safety_propagation_checks)
all_positive_checks.update(cross_agent_checks)
all_positive_checks.update(prohibited_action_checks)

passed = sum(bool(v) for v in all_positive_checks.values())
total = len(all_positive_checks)

orchestrator_status = (
    "WORKFLOW_COMPLETED_WITH_MANDATORY_SAFETY_BLOCK"
    if passed == total
    and safety_gate
        == "BLOCKED_INSUFFICIENT_FOR_CLINICAL_INTERPRETATION"
    else
    "WORKFLOW_COMPLETED"
    if passed == total
    else
    "REVIEW_REQUIRED"
)

# -----------------------------------------------------------------------------
# 8. FINAL SHARED WORKFLOW STATE
# -----------------------------------------------------------------------------

workflow_id = (
    f"NB9-{datetime.now(timezone.utc).strftime('%Y%m%dT%H%M%SZ')}"
)

shared_state = {
    "workflow_id": workflow_id,
    "timestamp_utc": datetime.now(timezone.utc).isoformat(),

    "system": {
        "name": "Automated Trustworthy Multi-Agent Healthcare AI Demonstrator",
        "stage": "NB9",
        "research_only": True,
    },

    "execution_order": expected_agent_order,

    "agent_execution": {
        agent: {
            "executed": True,
            "artifact_path": path,
        }
        for agent, path in expected_artifacts
    },

    "prediction_state": {
        "probability": (
            float(prediction_probability)
            if prediction_probability is not None
            else None
        ),
        "threshold": float(prediction_threshold),
        "predicted_class": (
            int(prediction_class)
            if prediction_class is not None
            else None
        ),
    },

    "safety_state": {
        "gate": safety_gate,
        "clinical_interpretation_allowed": False,
        "clinical_action_allowed": False,
        "safety_override_allowed": False,
        "mandatory": True,
    },

    "reporting_state": {
        "mode": cds_mode,
        "clinical_interpretation_allowed": False,
    },

    "governance": {
        "model_modification_allowed": False,
        "threshold_modification_allowed": False,
        "missing_value_fabrication_allowed": False,
        "manual_patient_imputation_allowed": False,
        "clinical_measurement_invention_allowed": False,
        "diagnosis_allowed": False,
        "treatment_recommendation_allowed": False,
        "safety_override_allowed": False,
        "causal_claims_allowed": False,
        "unsupported_shap_claims_allowed": False,
        "individual_fairness_claims_allowed": False,
    },

    "final_status": orchestrator_status,
}

# -----------------------------------------------------------------------------
# SAVE ORCHESTRATOR RESULT
# -----------------------------------------------------------------------------

orchestrator_result = {
    "agent": "Agentic Orchestrator",
    "agent_version": "NB9.1",
    "timestamp_utc": datetime.now(timezone.utc).isoformat(),

    "workflow_id": workflow_id,

    "execution_order": expected_agent_order,

    "integrity_checks": all_positive_checks,

    "audit_summary": {
        "checks_passed": passed,
        "checks_total": total,
        "all_checks_passed": passed == total,
    },

    "safety": {
        "mandatory_control_point": True,
        "safety_gate": safety_gate,
        "clinical_interpretation_allowed": False,
        "clinical_action_allowed": False,
        "safety_override_allowed": False,
    },

    "final_status": orchestrator_status,

    "shared_state": shared_state,

    "provenance": {
        agent: path
        for agent, path in expected_artifacts
    },
}

result_path = os.path.join(
    OUTPUT_DIR,
    "nb9_agentic_orchestrator_result.json"
)

with open(result_path, "w") as f:
    json.dump(orchestrator_result, f, indent=2)

state_path = os.path.join(
    OUTPUT_DIR,
    "nb9_final_shared_workflow_state.json"
)

with open(state_path, "w") as f:
    json.dump(shared_state, f, indent=2)

audit_path = os.path.join(
    TABLE_DIR,
    "nb9_agentic_orchestrator_audit.json"
)

with open(audit_path, "w") as f:
    json.dump(
        {
            "workflow_id": workflow_id,
            "agent_order": expected_agent_order,
            "checks_passed": passed,
            "checks_total": total,
            "all_checks_passed": passed == total,
            "safety_gate": safety_gate,
            "final_status": orchestrator_status,
            "integrity_checks": all_positive_checks,
        },
        f,
        indent=2
    )

# -----------------------------------------------------------------------------
# HUMAN-READABLE ORCHESTRATOR REPORT
# -----------------------------------------------------------------------------

report_path = os.path.join(
    REPORT_DIR,
    "nb9_agentic_orchestrator_summary.txt"
)

with open(report_path, "w") as f:
    f.write("NB9 AGENTIC ORCHESTRATOR\n")
    f.write("=" * 80 + "\n\n")

    f.write(f"Workflow ID: {workflow_id}\n")
    f.write(f"Research only: True\n\n")

    f.write("EXECUTION ORDER\n")
    f.write("-" * 80 + "\n")

    for i, agent in enumerate(expected_agent_order, start=1):
        f.write(f"{i}. {agent}\n")

    f.write("\nSAFETY STATE\n")
    f.write("-" * 80 + "\n")
    f.write(f"Safety gate: {safety_gate}\n")
    f.write("Clinical interpretation allowed: False\n")
    f.write("Clinical action allowed: False\n")
    f.write("Safety override allowed: False\n")

    f.write("\nFINAL ORCHESTRATOR STATUS\n")
    f.write("-" * 80 + "\n")
    f.write(f"{orchestrator_status}\n")

# -----------------------------------------------------------------------------
# CONSOLE OUTPUT
# -----------------------------------------------------------------------------

print("\n" + "-" * 80)
print("ORCHESTRATOR EXECUTION ORDER")
print("-" * 80)

for i, agent in enumerate(expected_agent_order, start=1):
    print(f"  {i}. {agent}")

print("\n" + "-" * 80)
print("ORCHESTRATOR STATE")
print("-" * 80)

print(f"Workflow ID                    : {workflow_id}")
print(f"Research only                  : True")
print(f"Input quality                  : {quality_status}")
print(f"Prediction probability        : {prediction_probability}")
print(f"Locked threshold              : {prediction_threshold}")
print(f"Predicted class               : {prediction_class}")
print(f"Safety gate                   : {safety_gate}")
print(f"Clinical interpretation       : False")
print(f"Clinical action               : False")
print(f"Safety override               : False")
print(f"CDS report mode               : {cds_mode}")

print("\n" + "-" * 80)
print("ORCHESTRATOR AUDIT")
print("-" * 80)

for key, value in all_positive_checks.items():
    print(f"  {key:<45}: {value}")

print(f"\nIntegrity checks passed        : {passed}/{total}")
print(f"Final orchestrator status     : {orchestrator_status}")

print("\n" + "=" * 80)
print(
    "CELL 20 RESULT: "
    f"{'PASSED' if passed == total else 'REVIEW REQUIRED'}"
)
print("=" * 80)

print("\nSaved:")
print(result_path)
print(state_path)
print(audit_path)
print(report_path)
```
